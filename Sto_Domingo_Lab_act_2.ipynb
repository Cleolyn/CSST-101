#2. Basic Calculations


Problem:

* A student received the following grades:


Quiz = 85, Laboratory = 90, project =, Exam = 80


* The grading weights are:


Quiz = 20%, laboratry = 25%, Project = 25%, Exam = 30%



*  Task:


Calculate the student's weighted final grade.




```r
quiz_grade <- 85
lab_grade <- 90
project_grade <- 88
exam_grade <- 80

quiz_weight <- 0.20
lab_weight <- 0.25
project_weight <- 0.25
exam_weight <- 0.30

weighted_quiz <- quiz_grade * quiz_weight
weighted_lab <- lab_grade * lab_weight
weighted_project <- project_grade * project_weight
weighted_exam <- exam_grade * exam_weight

final_grade <- weighted_quiz + weighted_lab + weighted_project + weighted_exam

print(paste("The student's weighted final grade is:", final_grade))
```

Instead of repeatedly writing the same code, create a function called:

```r
classify_grade <- function(grade) {
  if (grade >= 90 && grade <= 100) {
    return("Excellent")
  } else if (grade >= 85 && grade <= 89) {
    return("Very Good")
  } else if (grade >= 80 && grade <= 84) {
    return("Good")
  } else if (grade >= 75 && grade <= 79) {
    return("Passed")
  } else {
    return("Failed")
  }
}


print(classify_grade(92))
print(classify_grade(86))
print(classify_grade(81))
print(classify_grade(77))
print(classify_grade(65))
```

```r
classify_risk <- function(attendance, grade) {
  if (attendance < 75 && grade < 75) {
    return("High Risk")
  } else if (attendance < 85 || grade < 80) {
    return("Moderate Risk")
  } else {
    return("Low Risk")
  }
}

students_df$Risk <- mapply(classify_risk, students_df$Attendance, students_df$Grade)

print(students_df)
```

#14. Add a Recommendation

Extend your rule-based system.

For Example:

  Risk,      Recommendation

  High Risk,  Immediate academic intervention

  Moderate Risk,  Academic monitoring
  
  Low Risk,  Continue regular monitoring  


```r
get_recommendation <- function(risk_level) {
  if (risk_level == "High Risk") {
    return("Immediate academic intervention")
  } else if (risk_level == "Moderate Risk") {
    return("Academic monitoring")
  } else if (risk_level == "Low Risk") {
    return("Continue regular monitoring")
  } else {
    return("N/A")
  }
}

students_df$Recommendation <- sapply(students_df$Risk, get_recommendation)

print(students_df)
```
